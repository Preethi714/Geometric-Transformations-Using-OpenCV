3. Display the following using a **1 × 3 layout** in Matplotlib:

   - **Vertically Sheared Image**


```python
M1 = np.float32([[1, 0.5, 0], [0, 1, 0]])
M2 = np.float32([[1, 0, 0], [0.5, 1, 0]])
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
M1 = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1)
rotated45 = cv2.warpAffine(image, M1, (cols, rows))
```

```python
M2 = cv2.getRotationMatrix2D((cols/2, rows/2), 90, 1)
rotated90 = cv2.warpAffine(image, M2, (cols, rows))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(rotated45, cv2.COLOR_BGR2RGB))
plt.title("45° Rotated Image")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(rotated90, cv2.COLOR_BGR2RGB))
plt.title("90° Rotated Image")
plt.axis('off')

plt.show()
```
